```python
total_jobs = df['job_id'].nunique()

print("Total Jobs Posted:", total_jobs)
```

```python
jobs_by_city = df.groupby('city')['job_id'].nunique().sort_values(ascending=False)

print(jobs_by_city)
```

```python
jobs_by_skill = df.groupby('skill')['job_id'].nunique().sort_values(ascending=False)

print(jobs_by_skill)
```

```python
jobs_by_company = df.groupby('company_id')['job_id'].nunique().sort_values(ascending=False)

print(jobs_by_company)
```

```python
avg_threshold = df['skill_threshold'].mean()

print("Average Skill Threshold:", avg_threshold)
```

```python
max_threshold = df['skill_threshold'].max()

print("Highest Skill Threshold:", max_threshold)
```

```python
min_threshold = df['skill_threshold'].min()

print("Lowest Skill Threshold:", min_threshold)
```

```python
valid_thresholds = df[
    (df['skill_threshold'] >= 0) &
    (df['skill_threshold'] <= 100)
]

print("Valid Skill Thresholds:", len(valid_thresholds))
```

```python
invalid_thresholds = df[
    (df['skill_threshold'] < 0) |
    (df['skill_threshold'] > 100)
]

print("Invalid Skill Thresholds:", len(invalid_thresholds))
```

```python
df['event_timestamp'] = pd.to_datetime(df['event_timestamp'])
```

```python
print("Earliest Event:", df['event_timestamp'].min())
```

```python
print("Latest Event:", df['event_timestamp'].max())
```

```python
events_by_city = df.groupby('city')['event_id'].count().sort_values(ascending=False)

print(events_by_city)
```

```python
print("===== JOB SUPPLY SUMMARY =====")
print("Total Jobs Posted:", df['job_id'].nunique())
print("Unique Companies:", df['company_id'].nunique())
print("Unique Skills:", df['skill'].nunique())
print("Unique Cities:", df['city'].nunique())
print("Valid Skill Thresholds:", len(valid_thresholds))
print("Invalid Skill Thresholds:", len(invalid_thresholds))
print("Average Skill Threshold:", df['skill_threshold'].mean())
```

```python
print("Jobs by City:")
print(jobs_by_city)

print("\nJobs by Skill:")
print(jobs_by_skill)

print("\nJobs by Company:")
print(jobs_by_company)
```

```python
jobs_by_city.to_csv("jobs_by_city.csv")
jobs_by_skill.to_csv("jobs_by_skill.csv")
jobs_by_company.to_csv("jobs_by_company.csv")
```
